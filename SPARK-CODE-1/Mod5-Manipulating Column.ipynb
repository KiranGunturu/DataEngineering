# Dates and timestamp
# when we use current_date to get today's server date
    # date will be returned using yyyy-MM-dd format
# we can use current_timestamp to get current server time
    # timestamp will be returned using yyyy-MM-dd HH:mm:ss:SSS format
    # Hours will be by default 24 hrs format

```python
datetimes = [
    ("2014-02-28","2014-02-28 10:00:00.123"),
    ("2016-02-29","2016-02-09 08:08:08.999"),
    ("2017-10-31","2017-10-31 11:59:59.123"),
    ("2019-11-30","2019-11-30 00:00:00.000")
]
```

```python
dateframeDF=spark.createDataFrame(datetimes,schema="date string, time string")
```

```python
dateframeDF.show(truncate=False)
```

```python
# add 10 days to date and time values
# subtract 10 days to date and time values
```

```python
dateframeDF. \
withColumn("date_add_date",date_add("date",10)). \
withColumn("date_add_time",date_add("time",10)). \
withColumn("date_Sub_date",date_sub("date",10)). \
withColumn("date_sub_time",date_sub("time",10)).show(truncate=False)
```

```python
dateframeDF.\
withColumn("datediffdate",datediff(current_date(),"date")).show()
```

```python
dateframeDF.show()
```

```python
dateframeDF.\
withColumn("months_bw_date",round(months_between(current_date(),"date"),2)).\
withColumn("months_bw_time",round(months_between(current_timestamp(),"date"),2)).\
withColumn("add_months_to_date",add_months(current_date(),3)).\
withColumn("add_months_to_time",add_months(current_timestamp(),3)).show()
```

```python
help(trunc)
```

```python
# MM - first date of the month
# YY - first date of the year
# diff bw trunc and date_trunc is with trunc(date and format) and date_trunc(format and date)
dateframeDF.\
withColumn("month_trunc",trunc("date","MM")).\
withColumn("year_trunc",trunc("time","yy")).\
withColumn("week_trunc",trunc(current_date(),"week")).\
withColumn("quarter_trunc",trunc(current_date(),"quarter")).show()
```

```python
help(date_trunc)
```

```python
#date_trunc will add timestamp add zeros and not take from source column
dateframeDF.\
withColumn("month_trunc",date_trunc("MM","date")).\
withColumn("year_trunc",date_trunc("yy","time")).\
withColumn("week_trunc",date_trunc("week",current_date())).\
withColumn("quarter_trunc",date_trunc("quarter",current_date())).\
withColumn("first_time",date_trunc("HOUR","time")).show(truncate=False)
```
