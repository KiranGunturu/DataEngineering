```
schema = 'id int, name string, dop string, phone long, amount string, discount string'

data_list = [(100, "Prashant", "2020-06-15", 9238614990, "12000", 18.5),
             (101, "David", "2018-08-7", 8908617610, "15000", "nil"),
             (102, "Simran", "14-05-2019", None, "3000000000", 21)]
```

```
sales_df = spark.createDataFrame(data_list, schema)
```

```
sales_df.printSchema()
```

```
display(sales_df)
```

```
sales_df.select("dop", "amount", "discount").describe().show()
```

```
sales_df.selectExpr("string(id)", 
                       "cast(id as string)",
                       "cast(amount as int)") \
           .show()
```

```
final_sales_df = sales_df.selectExpr("id::string as transaction_id",
                                     "name as customer_name",
                                     "nvl(dop::date, to_date(dop,'d-M-y')) as date_of_purchase",
                                     "phone::string as customer_phone",
                                     "amount::int as purchase_amount",
                                     "nvl(discount::double, 0.0) as applied_discount") \
                        .filter("purchase_amount is not null")

final_sales_df.printSchema()
final_sales_df.show()
```
