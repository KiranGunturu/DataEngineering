# Drop records when all column values are nulls
# drop records when any of the column value is null
# drop records that have less than thresh non-null values
# drop records when any of the column value or all column values are nulls for provided subset of columns
# we can use df.na.drop or df.dropna to take care of delaing with records havving column with null values

# drop -> users_df.dropna
# fill -> users_df.fillna
# replace -> users_df.replacena

```python
employee= [
    (1,'scott',"Tiger",1000.0,10,True,"US"),
    (5,'sco',"Will",1250.0,10,True,"US"),
    (None,None,None,None,None,None,None),
    (None,None,None,None,None,None,None),
    (5,'sco',"Will",1250.0,10,True,"US"),
    (2,'James',"Micky",1250.0,None,True,"UK"),
    (3,'Mike',"Wilson",1400.0,'',False,"india"),
    (4,'Thomos',"George",1000.0,10,False,"AUS"),
    (6,'Tho',"Ge",1500.0,10,False,None),
    (7,'krish',None,None,None,None,None)
]
```

```python
df2=spark.createDataFrame(employee,schema="id int, fname string, lname string, salary float, bonus string, is_customer string,nation string")
```

```python
df2.show()
```

```python
df2.select('*').\
dropna().show()
```

```python
# drop only when all values in a row is null
df2.select('*').dropna('all').show()
```

```python
# drop if any column value is null
df2.select('*').dropna('any').show()
```

```python
df2.show()
```

```python
# drop if rows contain less than 3 non null values
df2.dropna(thresh=3).show()
```

```python
df2.dropna('all',subset=['bonus','id']).show()
```
