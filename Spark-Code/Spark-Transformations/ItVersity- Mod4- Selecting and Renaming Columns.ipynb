# Renaming column
# we can rename column or expression using alias as part of select
# we can add or rename column or expression using withColumn on top of dataframe
# we can rename one column at a time using withColumnRenamed on top of dataframe
# we typically use withColumn to perform row level transformations and then to provide a name to the result. if we provide the same name as existing  column, then the column will be replaced with new one
# if we want to just rename the column , then user withColumnRenamed
# if we want to apply any transformation, we need to either use select or withColumn
# we can rename bunch of columns use toDF and to change the order of columns use select

```python
user_df.select('id','name','lastname',concat(col('name'),lit(', '),col('lastname')).alias('fullname')).show()
```

```python
from pyspark.sql.functions import *
```

```python
user_df.select('id','name','lastname').withColumn('fullname',concat(col('name'),lit(', '),col('lastname'))) \
.drop('name','lastname').show()
```

```python
user_df.select('id','name','lastname').withColumn('ln',col('lastname')).show()
user_df.select('id','name','lastname').withColumn('ln','name').show() # this will fail
```

```python
user_df.select('id','name','courses').show()
```

```python
user_df.select('id','name','courses').withColumn('course_count',size('courses')).show()
```

```python
user_df.select('id','name','lastname').\
withColumnRenamed('id','user_id').\
withColumnRenamed('name','first_name').show()
```

# how to rename multiple columns at one go.

```python
required_columns=['id','name','lastname']
target_columns=['user_id','my_first_name','my_last_name']
```

```python
user_df.select(required_columns).show()
```

```python
user_df.select(required_columns).toDF(*target_columns).show()
# here the * represents the varying no of args and it will convert the list to tuple object
```
