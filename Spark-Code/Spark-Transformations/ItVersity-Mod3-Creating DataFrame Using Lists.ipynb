# special types that are supported in spark
# ARRAY , STRUCT, MAP

```python
# python stractures like dict and list can be explicitely converted to ARRAY AND MAP respectively
# we need to use few spark related APIs to convert data stractures STRUCT type
```

```python
my_array=[
    {'id':10,'name':'Kiran','phone':['603-218-9086','123-686-6786']},
    {'id':10,'name':'Pavani','phone':['604-218-9086','124-686-6786']},
    {'id':10,'name':'Krish','phone':['605-218-9086','125-686-6786']},
    {'id':10,'name':'James','phone':None},
    {'id':10,'name':'Murali','phone':['606-218-9086','126-686-6786']}
]
```

```python
df=spark.createDataFrame(Row(**user) for user in my_array)
```

```python
df.show(truncate=False)
```

```python
df.dtypes
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import explode
```

```python
df.withColumn('phonenumber',explode('phone')).drop('phone').show(truncate=False) 
# we will not get if there are any null values in array column for that we need to use explode_outer
```

```python
from pyspark.sql.functions import explode_outer
```

```python
df.withColumn('phonenumber',explode_outer('phone')).drop('phone').show(truncate=False) 
```

```python
df.show(truncate=False)
```

# if we want to devide the array columns into seperate column then follow below

```python
from pyspark.sql.functions import col
```

```python
df.select('id','phone',col('phone')[0].alias('HomePhone'),col('phone')[1].alias('Mobile')).show(truncate=False)
```

# MAP

```python
my_map=[
    {'id':10,'name':'Kiran','phone':{'Mobile':'603-218-9086','Home':'123-686-6786'}},
    {'id':10,'name':'Pavani','phone':{'Mobile':'604-218-9086','Home':'124-686-6786'}},
    {'id':10,'name':'Krish','phone':{'Mobile':'605-218-9086','Home':'125-686-6786'}},
    {'id':10,'name':'James','phone':None},
    {'id':10,'name':'Murali','phone':{'Mobile':'606-218-9086','Home':'126-686-6786'}}
]
```

```python
my_map
```

```python
h=spark.createDataFrame(Row(**user) for user in my_map)
```

```python
h.show(truncate=False)
```

```python
h.printSchema()
```

```python
h.select('id','name',col('phone')['Home']).show()
```

```python
h.select('id','name',col('phone')['Mobile']).show()
```

```python
h.dtypes
```

```python
h.select('id','name',col('phone')['Home'].alias('HomeNumber'),col('phone')['Mobile'].alias('HomeNumber')).show()
```

```python
h.select('id','name',explode('phone')).show()
```

```python
h.select('id','name',explode_outer('phone')).show()
```

```python
h.select('id','name',explode('phone')).\
withColumnRenamed('key','PhoneType').\
withColumnRenamed('value','Phone_Number').\
drop('phone').show()
```

# STRUCT

```python
my_struct=[
    {'id':10,'name':'Kiran','phone':Row(Mobile='603-218-9086',Home='123-686-6786')},
    {'id':10,'name':'Pavani','phone':Row(Mobile='604-218-9086',Home='124-686-6786')},
    {'id':10,'name':'Krish','phone':Row(Mobile='605-218-9086',Home='125-686-6786')},
    {'id':10,'name':'James','phone':Row(Mobile=None,Home=None)},
    {'id':10,'name':'Murali','phone':Row(Mobile='606-218-9086',Home='126-686-6786')}
]
```

```python
j=spark.createDataFrame(Row(**user) for user in my_struct)
```

```python
j.show(truncate=False)
```

```python
j.dtypes
```

```python
j.select('id','name','phone.Mobile','phone.Home').show(truncate=False)
```

```python
j.select('id','name','phone.*').show(truncate=False)
```
