```
data_list1 = [(100, "Prashant", "Software"),
             (101, "David", None),
             (102, "Sushant",None),
             (103, "Abdul", "Account"),
             (104, "Shruti", "Software")]

data_list2 = [(501,"Software"),
             (502, "Account")]

employee_df = spark.createDataFrame(data_list1).toDF("id", "first_name", "department")
department_df = spark.createDataFrame(data_list2).toDF("id", "department")
```

```
employee_df.join(department_df, "department", "left").select("first_name", "department").show()
```

```
person_list = [(100, "Prashant", 30),
             (101, "David", None),
             (102, "Sushant",None),
             (103, "Abdul", 45),
             (104, "Shruti", 28)]
person_df = spark.createDataFrame(person_list).toDF("id", "name", "age")
display(person_df)
```

```
person_df.select("*").where("age is null").show()
```

```
person_df.selectExpr("*", "age>29").show()
```

```
person_df.select("*").where("age>29").show()
```

```
person_df.selectExpr("avg(age)").show()
```

```
person_df.filter("age is not null").selectExpr("avg(age)").show()
```
