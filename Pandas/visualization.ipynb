```python
div=['Div-A','Div-B','Div-C','Div-D','Div-E']
```

```python
Marks=[70,82,72,65,69]
```

```python
import matplotlib.pyplot as plt
```

```python
# Vertical Bar
fig=plt.figure(figsize=(5,5))
plt.bar(div, Marks,color='green')
plt.title('Avg Marks by Division')
plt.xlabel('Divisons')
plt.ylabel('Marks')
```

```python
# Horizontal Bar
fig=plt.figure(figsize=(5,5))
plt.barh(div, Marks,color='green')
plt.title('Avg Marks by Division')
plt.xlabel('Divisons')
plt.ylabel('Marks')
plt.grid(True)
plt.savefig('Horizontal Bar')
```

```python
# Side by Side bar
import numpy as np
data1=[23,85,72,43,25]
data2=[42,35,41,16,9]
width=0.25
plt.bar(np.arange(len(data1)),data1,color='red',width=width)
plt.bar(np.arange(len(data2))+width,data2,color='black',width=width)
```

```python
# Stacked Bar
data1=[23,85,72,43,25]
data2=[42,35,41,16,9]
width=0.25
plt.bar(np.arange(len(data1)),data1,color='black',width=width)
plt.bar(np.arange(len(data2)),data2,color='red',bottom=data1,width=width)
```

```python
# Line Chart
x=[2,4,6]
y=[1,3,5]

fig=plt.figure(figsize=(5,5))
plt.plot(x,y)
plt.title('Line Chart')
```

```python
year=[2010,2011,2012,2013,2014,2015]
sales=[4000,5000,8000,10000,20000,500]
profit=[5000,7000,2000,500,100,450]
discount=[5,9,20,15,0,11]
# - solid
# -- dashed
# -. dash-dotted
# : dotted line
plt.plot(year,sales,color='green',linewidth=2.0,linestyle='-')
plt.plot(year,profit,color='orange',linewidth=2.0,linestyle=':')
plt.plot(year,discount,color='blue',linewidth=2.0,linestyle=':')
```

```python
# Area Chart
fig=plt.figure(figsize=(10,5))
x = range(1,15)
y = [2,3,5,5,6,7,8,9,6,7,8,9,2,4]
plt.fill_between(x,y,color='skyblue',alpha=0.5) # Alpha is reduce rhe color brightness
plt.plot(x,y,color='blue',alpha=0.9) # to add boarder
```

```python
# Stacked Area Chart
##### Legends Location #######
    #Location String	Location String
    #Best	0
    #Upper right	1
    #Upper left	2
    #Lower left	3
    #Lower right	4
    #Right	5
    #Center left	6
    #Center right	7
    #Lower center	8
    #Upper center	9
    #center	10
    
x=range(1,6)
x
y=[[4,5,6,7,8],[6,7,8,3,8],[4,6,7,3,9]]
plt.stackplot(x,y,labels=['A','B','C'])
plt.legend(loc=1)
```

```python
# Pie Chart
fig=plt.figure(figsize=(10,5))
firm_Name=['Firm-A','Firm-B','Firm-C','Firm-D','Firm-E']
Market_Share=[20,25,15,10,30]
explode=[0,0.1,0,0,0]
plt.pie(Market_Share,labels=firm_Name,shadow=True,explode=explode,autopct='%1.1f%%')
#autopct is to dispaly the percentages
# 1.1 means it will give one decimal point and 1.2 means will get two decimal points
plt.legend(title='Firm_Name',loc=1)
```
