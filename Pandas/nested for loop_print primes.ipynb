```python
def count_no_of_Chars(s):
    count ={}
    for i in s:
        if i in count:
            count[i]+=1
        else:
            count[i]=1
    return count

s = ['Kiran','Kiran','Hello']
count_no_of_Chars(s)
```

```python
def print_items(n):
    for i in range(n):
        for j in range(n):
            for k in range(n):
                print(i,j,k)
print_items(3)
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
factorial(4)
```
