```python
from math import sqrt
def isPrime(n):
    if n<=1:
        return False
    for i in range(2,int(sqrt(n))):
        if n % i == 0:
            return False
    return True

ans=isPrime(14)
print(ans)
```

```python
from math import sqrt
def isPrime(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True

def PrintPrimes(n):
    for i in range(2,n+1):
        if isPrime(i):
            print(i,end=" ")

PrintPrimes(40)
```
