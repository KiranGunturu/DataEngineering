```python
def dictdf(data_dict1,data_dict2):
    df1=pd.DataFrame(data_dict1)
    df2=pd.DataFrame(data_dict2)
    cdf=pd.concat([df1,df2],ignore_index=True)
    return cdf
data_dict1={"one":["X","A"],"Two":["Y","B"]}
data_dict2={"one":["C","D"],"Two":["E","F"]}
dictdf(data_dict1,data_dict2)
```
