```
airlines_df = spark.read \
                .format("csv") \
                .option("header", "true") \
                .option("inferSchema","true") \
                .option("samplingRatio", "0.0001") \
                .load("/databricks-datasets/airlines/part-00000")
```

```
display(airlines_df)
```

```
summary_df = airlines_df.groupBy("UniqueCarrier", "Origin", "Dest", "FlightNum").count()
display(summary_df)
```

```
summary_df.orderBy("count").show()
```

```
from pyspark.sql.functions import expr
summary_df.orderBy(expr("count desc")).show()
```

```
summary_df.orderBy("count", ascending=False).show()
```

```
from pyspark.sql.functions import desc
summary_df.orderBy(desc("count")).show()
```

```
from pyspark.sql.functions import col
summary_df.orderBy(col("count").desc()).show()
```

```
from pyspark.sql.functions import desc
summary_df.orderBy("UniqueCarrier", "Origin", "Dest", desc("count")).show()
```

```
summary_df.orderBy(["UniqueCarrier", "Origin", "Dest", "count"], ascending=[1,1,1,0]).show()
```

```
from pyspark.sql.functions import desc
summary_df.orderBy("UniqueCarrier", "Origin", "Dest", desc("count")).limit(5).show()
```

```
from pyspark.sql.functions import sum, desc
airlines_df.groupBy("FlightNum") \
            .agg(sum("Distance").alias("total_distance")) \
            .sort(desc("total_distance")) \
            .limit(5) \
            .show()
```
