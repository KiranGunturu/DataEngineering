```python
import numpy as np
type(np.nan) # float
```

```python
# if we have a column data type as float and it also have null values in it
# we can't convert that float to int
# instead we should fill that null value then convert to int
df['sales'].fillna(0).astype('int')
df['sales'].isna() # it will give true or false
df[df['sales'].isna()] # it will give complete rows where we have sales column has null values
```
