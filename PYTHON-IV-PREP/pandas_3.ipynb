```python
import pandas as pd
df=pd.read_csv("C:/Users/X071854/Desktop/BigData/Pandas/orders.txt")
```

```python
#get me distict values of category column
df['category'].unique()
```

```python
# get me counts for each category like group by
df['category'].value_counts()
```

```python
# get me counts for each category like group by (multiple columns)
df[['category','city']].value_counts()
```

```python
fc = (df.category == 'Furniture')
df[fc]
```

```python
# multiple conditions with and
fc = (df.category == 'Furniture') & (df.sales > 1500)
df[fc]
```

```python
# multiple conditions with or
fc = (df.category == 'Furniture') | (df.category == 'Technology')
df[fc]
```

```python
# multiple values as IN operator instead using or condition multiple times
cate=['Furniture','Technology']
fc = df['category'].isin(cate)
df[fc]
# show only passed columns
df[fc]['order_id']['category']
```

```python
#write to csv
df.to_csv('orders_furniture.csv')
```
