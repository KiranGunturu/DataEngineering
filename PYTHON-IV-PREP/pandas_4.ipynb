```python
#aggs
import pandas as pd
df = pd.read_csv('orders.txt')
df.head()
```

```python
# aggregations
# if we select one column, then it is a series object
df['sales'].sum()
df['sales'].avg()
df['sales'].max()
df['sales'].min()
```

```python
# get me the sum of sales for Furniture Category
fc = (df['category'] == 'Furniture')
df[fc]['sales'].sum()
```

```python
# get me sum of sales for all categories
# this wll give us the series
df.groupby('category')['sales'].sum()
```

```python
# convert series obj to df
df_cat_sales = df.groupby('category')['sales'].sum()
df_cat_sales.index
df_cat_sales.values
pd.DataFrame({'category':df_cat_sales.index, 'sales':df_cat_sales.values})
```

```python
# get me sum of sales for all categories
# this wll give us df object
df.groupby('category', as_index = False)['sales'].sum()
```

```python
# grouping on multiple columns and take sum of sales (only one agg)
df.groupby(['category''city'], as_index = False)['sales'].sum()
```

```python
# grouping on multiple columns and take sum of sales and profit (multiple aggs)
df.groupby(['category''city'], as_index = False)['sales'].agg({'sales':'sum','profit':'sum'})
```

```python
#merge #join
dfr = pd.read_csv('returns.txt')
dfr
```

```python
# inner join
pd.merge(left = df, right = dfr, how = 'inner', left_on = 'order_id', 'right_on' = 'order_id') 
```

```python
# left join
pd.merge(left = df, right = dfr, how = 'left', left_on = 'order_id', 'right_on' = 'order_id') 
```

```python
# right join
pd.merge(left = df, right = dfr, how = 'right', left_on = 'order_id', 'right_on' = 'order_id')
```

```python
# outer join
pd.merge(left = df, right = dfr, how = 'outer', left_on = 'order_id', 'right_on' = 'order_id')
```

```python
# cross join
pd.merge(left = df, right = dfr, how = 'cross')
```

```python
# join on multiple columns
pd.merge(left = df, right = dfr, how = 'right', left_on = ['order_id','product_id'], 'right_on' = ['order_id','product_id'])
```

```python
# drop duplicates
emp = {'name':['Kiran','Mary','John','Kiran'],'gender':['male','female','male','male'],'email':['abc@gmail.com','def@gmail.com','ghi@gmail.com','abc@gmail.com']}
df1 = pd.DataFrame(emp)
```

```python
df1.drop_duplicates() # when full row is duplicates
```

```python
df.drop_duplicates(subset = 'email_id') # delete if there are dupes in email_id, by default it will keep the first value and delete the last value
```

```python
df.drop_duplicates(subset = 'email_id', keep = 'last')  it will now delete the first value value and keep last value
```

```python
df.drop_duplicates(subset = ['email_id','gender'], keep = 'last')  it will now delete the first value value and keep last value
```

```python
df.drop_duplicates(subset = 'email_id', keep = False) # it will delete all dupe rows on email_id. it wont keep either last or first
```

```python
# handle nulls
import pandas as pd
import numpy as np
emp = {'name':['Kiran','Mary','John','Kiran'],'gender':[np.nan,'female','male','male'],'email':['abc@gmail.com','def@gmail.com','ghi@gmail.com',np.nan]}
df1 = pd.DataFrame(emp)
```

```python
df1.isna()
```

```python
df1['gender'].isna()
```

```python
df1.dropna(how='any') # if any col is null then it will drop the full row
```

```python
df1.dropna(how='all') # drop a row when all values are null in row
```

```python
# drop a column based on nulls
df1.dropna(how='all', axis='columns') # it will drop a column when all values of column is null
df1.dropna(how='any', axis='columns') # it will drop a column when at least one value of column is null
```

```python
# fill nulls
df1.fillna(0) # fill all null values to 0
```

```python
df1['gender'].fillna(value='new_male') # it will replace all null values of gender column to new_male string
```
