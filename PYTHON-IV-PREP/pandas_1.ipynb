```python
# rows from 1 to 3 (3 will not be considered but till 2)
df_emp.iloc[1:3]
```

```python
# rows between 1 and 3 (3 will not be considered but till 2) and 0 to 2 columns (0th and 1st index as 2 will not be considered)
df_emp.iloc[1:3,0:2]
```

```python
#what if we want random rows like 0th and 2nd row along with 1st and 2nd column
df_emp.iloc[[0,2],[1,2]]
```

```python
# loc - label based
df_emp.loc[1]
```

```python
df_emp.loc[1,'name']
```

```python
# here lable 2 is also included in the result. where as with iloc, last index will be ignored as iloc is more on index based
# give all rows from 0 to 2 and give me only name and email columns
df_emp.loc[0:2,['name','email']]
```

```python
# give all rows from 0 to 2 and give all the columns from name to email
df_emp.loc[0:2,'name':'email']
```
